This notebook takes the persons.csv created, and adds a column with names of researchers written in a way that is required during search in DRP

```python
import pandas as pd

# Load CSV file
df = pd.read_csv('persons.csv')

df['first_name'] = df['first_name'].str.strip()
df['last_name'] = df['last_name'].str.strip().str.replace(r'\s*-\s*$', '', regex=True)

df['combined'] = df['last_name'] + ', ' + df['first_name']

# Show first 5 rows
df.head()
```

```python
df.to_csv('persons_compatible.csv', index=False)
```
