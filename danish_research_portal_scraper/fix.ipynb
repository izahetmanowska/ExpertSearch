```python
import pandas as pd

df = pd.read_csv("danish_research_all.csv", sep=";", quotechar='"', encoding="utf-8")

df.columns

# Drop specific columns
df = df.drop(columns=["source_file", "review", "level", "publisher", "url", "affiliations", "source"])

df.head
```

```python
print("open_access_url:")
print(df["open_access_url"].notna().sum(), "present")
print(df["open_access_url"].isna().sum(), "missing")

print("\ndoi:")
print(df["doi"].notna().sum(), "present")
print(df["doi"].isna().sum(), "missing")

both_missing = df["open_access_url"].isna() & df["doi"].isna()
print("\nRows where both are missing:", both_missing.sum())

both_present = df["open_access_url"].notna() & df["doi"].notna()
print("\nRows where both are present:", both_present.sum())

print("\n A DOI, or Digital Object Identifier, is a string of numbers, letters and symbols used to uniquely identify an article or document, and to provide it with a permanent web address (URL). A DOI is a permanent address for an article you're citing — it will always refer to that article, and only that one.")
print("\n Take doi first, if doesnt exist take from open access url")
```

```python
# Fill missing DOI values with open_access_url
df["doi"] = df["doi"].fillna(df["open_access_url"])

# Drop the open_access_url column
df = df.drop(columns=["open_access_url", "journal"])

print("Missing DOI after fill:", df["doi"].isna().sum())
```

```python
# Rename columns to fit paper.csv
df = df.rename(columns={
    "doi": "file",
    "author_count": "totalNumberOfContributors",
    "lang": "language",
    "id": "researchPortalId",
})
```

```python
# Change column order to fit paper.csv
order = [
    "uuid", "year", "title", "subtitle", "university",
    "type", "abstract", "file", "totalNumberOfContributors"
]

df = df[[c for c in order if c in df.columns] +
        [c for c in df.columns if c not in order]]
```

```python
# Drop the language column
df = df.drop(columns=["language"])
```

```python
df.to_csv("fixed.csv", index=False, encoding="utf-8")
```
