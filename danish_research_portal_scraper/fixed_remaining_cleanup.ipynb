```python
import pandas as pd

# Load CSV
df = pd.read_csv("fixed_remaining.csv")

# Treat empty strings as missing
df = df.replace(r'^\s*$', pd.NA, regex=True)

# Optional: normalize title spacing so near-identical titles match better
df["title"] = df["title"].astype(str).str.strip()

# Function to merge a group of duplicate titles into one row
def merge_group(group):
    result = {}
    for col in group.columns:
        # take the first non-empty value in that column
        non_null = group[col].dropna()
        result[col] = non_null.iloc[0] if not non_null.empty else pd.NA
    return pd.Series(result)

# Merge duplicates by title
deduped = (
    df.groupby("title", dropna=False, as_index=False)
      .apply(merge_group)
      .reset_index(drop=True)
)

# Save result
deduped.to_csv("DRP.csv", index=False)

print(f"Original rows: {len(df)}")
print(f"Rows after deduplication: {len(deduped)}")
```
